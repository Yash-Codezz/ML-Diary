---

```python
import numpy as np

import pandas as pd

from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
data = pd.read_csv("cell_diagnosis_svm.csv")
```

```python
data.drop(columns=['SampleID', 'LabReference', 'SampleAge'], inplace=True)
```

## Building Pipeline:

```python
X = data.drop(columns=['Diagnosis'])
y = data['Diagnosis']
```

```python
impute_pipeline = Pipeline(steps=[
    ('impute', SimpleImputer(strategy='median'))
])

encoder_pipeline = Pipeline(steps=[
    ('encode', OneHotEncoder(drop='first'))
])
```

```python
preprocessing = ColumnTransformer(transformers=[
    ('Impute', impute_pipeline, ['CellTexture', 'CellArea', 'Smoothness', 'DNAContent']),
    ('Encode', encoder_pipeline, ['TissueType'])
], remainder='passthrough')

X_transform = preprocessing.fit_transform(X)
```

```python
encoder = LabelEncoder()

y_transform = encoder.fit_transform(y)
```

```python
X_cols = preprocessing.get_feature_names_out()

X_df = pd.DataFrame(X_transform, columns=X_cols)

y_df = pd.DataFrame({'Diagnosis' : y_transform})

df = pd.concat([X_df, y_df], axis=1)
```

```python
df.to_csv("preprocessed_cell_diagnosis_svm.csv")
```
