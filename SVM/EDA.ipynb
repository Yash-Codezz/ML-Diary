```python
import pandas as pd

import matplotlib.pyplot as plt

import seaborn as sns
```

```python
data = pd.read_csv("cell_diagnosis_svm.csv")
```

```python
print(f"Rows: {data.shape[0]}\nColumns: {data.shape[1]}")
```

```python
print(f"Duplicated Values: {data.duplicated().sum()}")
```

```python
print(f"Missing Values(%):\n{((data.isna().sum() / len(data)).round(3) * 100).sort_values(ascending=False)}")
```

```python
data.info()
```

```python
data.describe()
```

```python
data.sample(10)
```

```python
print(data['Diagnosis'].value_counts())

figure = plt.figure(figsize=(7,5))

sns.countplot(data=data, x=data['Diagnosis'])
plt.show("Count-Plot")

plt.show()
```

```python
print(f"Categorical Columns: {data.select_dtypes(include=['object']).columns.tolist()}")
```

```python
pd.crosstab(index=data['TissueType'], columns=data['Diagnosis'], normalize='index').round(3) * 100
```

```python
figure = plt.figure(figsize=(7,5))

sns.countplot(data=data, x='TissueType', hue='Diagnosis')
plt.show("Count-Plot")

plt.show()
```

# Summary:

- **Target Feature:** *Diagnosis*.

- **Input Features (Numerical):**
| Features | Feature Engineering |
| - | - |
| SampleAge | Drop |
| CellTexture | SimpleImputer(median) |
| CellArea | SimpleImputer(median) |
| Smoothness | SimpleImputer(median) |
| DNAContent | SimpleImputer(median) |

- **Input Features (Categorical):**
| Features | Feature Engineering |
| - | - |
| TissueType | One-Hot-Encoder |

- There are no duplicated values.

- *SampleID* and *LabReference* are noisy features, hence *droped*.
