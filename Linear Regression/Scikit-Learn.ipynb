# Linear Regression using Scikit-Learn.

---

```python
import time

import numpy as np

import pandas as pd

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_percentage_error
```

```python
data = pd.read_csv("house_prices_preprocessed.csv")
```

```python
data.drop(columns=['Unnamed: 0'], inplace=True)
```

## Train-Test-Split:

```python
X = data.drop(columns=['SalePrice'])
y = data['SalePrice']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=69)
```

## Building Model:

```python
Lr = LinearRegression(n_jobs=-1)

start_time = time.time()
Lr.fit(X_train, y_train)
end_time = time.time() - start_time

print("Model Trained Successfully!")
print(f"Time Taken in training: {end_time:.2f}s")
```

## Result:

```python
y_pred = Lr.predict(X_test)

print(f"r2 Score: {(r2_score(y_test, y_pred)*100):.2f}%")
print(f"\nmean absolute error: {(mean_absolute_percentage_error(y_test, y_pred)*100):.2f}%")
```

```python
print(f"r2 Score: {np.mean(cross_val_score(estimator=Lr, X=X, y=y, scoring='r2', cv=10, n_jobs=-1)) * 100:.2f}%")
```

```python
print(f"mean absolute error: {-np.mean(cross_val_score(estimator=Lr, X=X, y=y, scoring='neg_mean_absolute_percentage_error', cv=10, n_jobs=-1)) * 100:.2f}%")
```
