# Feature Engineering:

---

```python
import numpy as np

import pandas as pd

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
data = pd.read_csv("house_prices.csv")
```

```python
data.drop(columns=['HouseID', 'InspectionCode', 'Floors', 'MonthlyHOAFee'], inplace=True)
```

## Building Pipeline:

```python
X = data.drop(columns=['SalePrice'])
y = data['SalePrice']
```

```python
X['Rooms'] = X['Bedrooms'] + X['Bathrooms']

X.drop(columns=['Bedrooms', 'Bathrooms'], inplace=True)
```

```python
num_impute = Pipeline(steps=[
    ('num_impute', SimpleImputer(strategy='median'))
])

garage_pipeline = Pipeline(steps=[
    ('garage_impute', SimpleImputer(strategy='most_frequent')),
    ('garage_encode', OneHotEncoder(drop='first'))
])

encode = Pipeline(steps=[
    ('encode_ohe', OneHotEncoder(drop='first'))
])
```

```python
preprocessing = ColumnTransformer(transformers=[
    ('transform_num', num_impute, ['Area', 'Age']),
    ('transform', garage_pipeline, ['Garage']),
    ('transform_cat', encode, ['Neighborhood', 'Condition', 'HasGarden'])
], remainder='passthrough')
```

## Extracting Final Data:

```python
X_transform = preprocessing.fit_transform(X)

X_cols = preprocessing.get_feature_names_out()

X_df = pd.DataFrame(X_transform, columns=X_cols)

new_df = pd.concat([X_df, y], axis=1)
```

```python
new_df.to_csv('house_prices_preprocessed.csv')
```
