# KNN Using Scikit Learn.

---

```python
import time

import numpy as np

import pandas as pd

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```python
data = pd.read_csv("preprocessed_customer_churn_knn.csv")
```

```python
data.drop(columns=['Unnamed: 0'], inplace=True)
```

## Train Test Split:

```python
X = data.drop(columns=['Churn'])
y = data['Churn']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=69)
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

## Building Model:

```python
param = {'n_neighbors' : range(1,11,2),
        'weights' : ['distance', 'uniform'],
        'p' : [1, 2]}

grid = GridSearchCV(
    estimator=KNeighborsClassifier(),
    param_grid=param,
    cv=5,
    scoring='recall',
    n_jobs=-1
)

grid.fit(X_train, y_train)

grid.best_params_
```

```python
knn = KNeighborsClassifier(n_neighbors=1, weights='distance', p=2)

start_time = time.time()
knn.fit(X_train, y_train)
end_time = time.time() - start_time

print("Model Trained Successfully!")
print(f"Time Taken for Training: {end_time:.2f}s")
```

## Results:

```python
y_pred = knn.predict(X_test)

print(f"Accuracy: {(accuracy_score(y_test, y_pred)*100):.2f}%")

print(f"\nPrecision: {(precision_score(y_test, y_pred)*100):.2f}%")

print(f"\nRecall: {(recall_score(y_test, y_pred)*100):.2f}%")

print(f"\nF1 Score: {(f1_score(y_test, y_pred)*100):.2f}%")

print("\n")
matrix = confusion_matrix(y_test, y_pred)
matrix_df = pd.DataFrame(matrix, index=['Actual Churn_No', 'Actual Churn_Yes'], columns=['Predicted Churn_No', 'Predicted Churn_Yes'])
matrix_df
```
