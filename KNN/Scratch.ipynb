```python
import time

import numpy as np

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```python
data = pd.read_csv("preprocessed_customer_churn_knn.csv")
```

```python
data.drop(columns=['Unnamed: 0'], inplace=True)
```

## Train-Test-Split:

```python
X = data.drop(columns=['Churn'])
y = data['Churn']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=69)
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

## Building Model:

```python
class KNeighborsClassifier:

    def __init__(self, n_neighbors):
        self.n_neighbors = n_neighbors

    def fit(self, X_train, y_train):
        self.X_train = np.array(X_train)
        self.y_train = np.array(y_train)

    def predict(self, X_test):

        predictions = []

        for i in range(X_test.shape[0]):
            distances=[]
            
            for j in range(self.X_train.shape[0]):
                distance = np.sqrt(np.sum((self.X_train[j] - X_test[i])**2))
                distances.append(distance)

            distances = np.array(distances)
            nearest_neighbors = np.argsort(distances)[:self.n_neighbors]

            # When dataset is imbalance we use weights:
            nearest_distance = distances[nearest_neighbors]
            weights = 1 / nearest_distance

            labels = self.y_train[nearest_neighbors]
            yes_weights = np.sum(weights[labels == 1])
            no_weights = np.sum(weights[labels == 0])
    
            prediction = 1 if(yes_weights > no_weights) else 0
    
            predictions.append(prediction)
            
        return np.array(predictions)
```

```python
class KNeighborsRegressor:

    def __init__(self, n_neighbors):
        self.n_neighbors = n_neighbors

    def fit(self, X_train, y_train):

        self.X_train = np.array(X_train)
        self.y_train = np.array(y_train)

    def predict(self, X_test):

        predictions = []
        
        for i in range(X_test.shape[0]):
            distances = []
            
            for j in range(self.X_train.shape[0]):
                distance = np.sqrt(np.sum(self.X_train[j] - X_test[i])**2)
                distances.append(distance)

            distances = np.array(distances)
            nearest_neighbors = np.argsort(distances)[:self.n_neighbors]

            values = self.y_train[nearest_neighbors]

            prediction = np.mean(values)

            predictions.append(prediction)

        return predictions
```

```python
knn = KNeighborsClassifier(n_neighbors=1)

start_time = time.time()
knn.fit(X_train, y_train)
end_time = time.time() - start_time

print("Model Trained Successfully!")
print(f"Time Taken for Training: {end_time:.2f}s")
```

## Results:

```python
y_pred = knn.predict(X_test)

print(f"Accuracy: {(accuracy_score(y_test, y_pred)*100):.2f}%")

print(f"\nPrecision: {(precision_score(y_test, y_pred)*100):.2f}%")

print(f"\nRecall: {(recall_score(y_test, y_pred)*100):.2f}%")

print(f"\nF1 Score: {(f1_score(y_test, y_pred)*100):.2f}%")

print("\n")
matrix = confusion_matrix(y_test, y_pred)
matrix_df = pd.DataFrame(matrix, index=['Actual Churn_No', 'Actual Churn_Yes'], columns=['Predicted Churn_No', 'Predicted Churn_Yes'])
matrix_df
```
