# Feature Engineering:

---

```python
import numpy as np

import pandas as pd

from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
data = pd.read_csv("customer_churn_knn.csv")
```

```python
data.drop(columns=['CustomerID', 'InternalReference', 'Age', 'MonthlyCharges'], inplace=True)
```

## Building Pipeline:

```python
X = data.drop(columns=['Churn'])
y = data['Churn']
```

```python
preprocessing = ColumnTransformer(transformers=[
    ('Impute_Mode', SimpleImputer(strategy='most_frequent'), ['SatisfactionScore']),
    ('Impute_Median', SimpleImputer(strategy='median'), ['DaysSinceLastLogin', 'TotalSpend']),
    ('Ordinal', OrdinalEncoder(categories=[['Basic', 'Standard', 'Premium']]), ['InternetPlan']),
    ('OneHot', OneHotEncoder(drop='first', sparse_output=False), ['HasFamilyPlan', 'AutoPayment', 'ContractType'])
], remainder='passthrough')

X_transform = preprocessing.fit_transform(X)

X_transform_cols = preprocessing.get_feature_names_out()
```

## Extracting Final Data:

```python
X_transform_df = pd.DataFrame(X_transform, columns=X_transform_cols, index=X.index)

y = pd.DataFrame({'Churn': y})

new_df = pd.concat([X_transform_df, y], axis=1)
```

```python
new_df.to_csv("preprocessed_customer_churn_knn.csv")
```
